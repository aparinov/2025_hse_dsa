```python
import pandas as pd
df = pd.read_excel('data/requests_data.xlsx')
print(len(df))
pd.set_option('display.max_columns', None)
df.head()


```

```python
important_cols = [
       'Номер заявки', 
       'Дата подачи заявки', 
       'Статус заявки',
       'Дата смены статуса заявки', 

       'Номер вакансии/темы',
       'Тема ВКР/КР',

       'Тип инициатора вакансии/темы', 
       
       'E-mail руководителя',

       'Почта студента', 
       'Кампус студента',
       'Факультет студента', 
       'ОП студента', 
       'Уровень образования',
       'Курс студента',
]
df[important_cols].sample(3).to_dict(orient='records')
```

```python
display(df['Номер вакансии/темы'].value_counts().value_counts())
display(df['Инициатор вакансии/темы'].value_counts().value_counts())
print(df['Дата подачи заявки'].min(), df['Дата подачи заявки'].max())
display(df['Почта студента'].value_counts().value_counts())
display(df['Тип инициатора вакансии/темы'].value_counts())
display(df['E-mail руководителя'].value_counts().value_counts().head(10))
```
